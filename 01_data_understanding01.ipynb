# 01 - Data Understanding
Load all CSV files and understand shape, columns and data types.

## Step 1: Import libraries

```python
import os
import pandas as pd

# show all columns when we print a table
pd.set_option("display.max_columns", None)
```

## Step 2: Set the folder where CSV files are stored

```python
# notebook is inside notebooks/ and CSV files are inside data/raw/
data_folder = "../data/raw"
```

## Step 3: Load all CSV files into one dictionary

```python
tables = {}   # key = table name, value = DataFrame

for file_name in sorted(os.listdir(data_folder)):
    if file_name.endswith(".csv"):
        table_name = file_name.replace(".csv", "")
        file_path = os.path.join(data_folder, file_name)
        tables[table_name] = pd.read_csv(file_path)

print("Total tables loaded:", len(tables))
for name in tables:
    print("-", name)
```

## Step 4: Shape of every table (rows, columns)

```python
for name, df in tables.items():
    print(name, "->", df.shape)
```

```python
# Same thing as a neat summary table
summary = pd.DataFrame({
    "table": list(tables.keys()),
    "rows": [df.shape[0] for df in tables.values()],
    "columns": [df.shape[1] for df in tables.values()]
})
summary
```

## Step 5: Column names of every table

```python
for name, df in tables.items():
    print(name)
    print(list(df.columns))
    print()
```

## Step 6: Data types of every table

```python
# Note: date columns usually show as 'object' or 'str' (text). We will fix them in the cleaning step.
for name, df in tables.items():
    print("TABLE:", name)
    print(df.dtypes)
    print("-" * 40)
```

## Step 7: Full info of every table (columns, non-null count, data type, memory)

```python
for name, df in tables.items():
    print("TABLE:", name)
    df.info()
    print("-" * 40)
```

## Step 8: First 5 rows of every table

```python
for name, df in tables.items():
    print("TABLE:", name)
    display(df.head())
    print()
```

## Step 9: Look at one table in detail
Change the table name below to any table from Step 3.

```python
table_name = list(tables.keys())[0]   # first table. Replace with e.g. "orders" if you want
df = tables[table_name]

print("Table :", table_name)
print("Shape :", df.shape)
display(df.head())
display(df.describe(include="all"))
```

## Step 10: Data dictionary (all tables, all columns)

```python
rows = []

for name, df in tables.items():
    for col in df.columns:
        rows.append({
            "table": name,
            "column": col,
            "data_type": str(df[col].dtype),
            "missing_values": df[col].isnull().sum(),
            "unique_values": df[col].nunique()
        })

data_dictionary = pd.DataFrame(rows)
data_dictionary
```
